```python
# Функция для вычисления Cramér’s V
def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = stats.chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    r, k = confusion_matrix.shape
    return np.sqrt(chi2 / (n * (min(r, k) - 1)))

# Оставляем только категориальные признаки
cat_features = autos_train2.select_dtypes(include=['object'])

# Строим матрицу Cramér’s V
cramers_matrix = pd.DataFrame(index=cat_features.columns, columns=cat_features.columns)

for col1 in cat_features.columns:
    for col2 in cat_features.columns:
        if col1 == col2:
            cramers_matrix.loc[col1, col2] = 1.0  # Диагональ = 1
        else:
            cramers_matrix.loc[col1, col2] = cramers_v(cat_features[col1], cat_features[col2])

# Приводим к числовому типу и округляем
cramers_matrix = cramers_matrix.astype(float).round(2)

# Выводим результат
print(cramers_matrix)
```

- описательную статистику,построили гистограммы и боксплоты для **числовых признаков**. Удаление выбросов и замена нереалистичных значений. 
 
Выводы по числовым признакам:
1) основная масса продаваемых атомобилей была зарегистрирована в 1995-2010 годах

2) большинство авто не очень мощные (60-170 л.с.)

3) основная часть машин с большим пробегом (140 тыс км)

4)  Нулевой месяц регистрации оставили как значение-заглушку. Если признак окажется важным можно попробовать заменить каким-то значением, но т.к. сезонности на гистограмме не прослеживалось, то скорее всего признак окажется неважным

5) В основном зарегистрированы машины с пятизначным индексом страны

6) большинство авто продают по невысокой цене. График распределения экпоненциально убывает

- графики распределения для **категориальных признаков** (барплоты и пайчарты)
Выводы по числовым признакам:

1) Наибольшие категории машин относятся к маленьким, седанам или универсалы c механической коробкой передач.

2) Механика встречается гораздо чаще (78,9%) чем автомат (19,1%)

3) практически все машины в качестве топлива используют бензин.

4) В датасете преобладают машины марки 'Фольксваген' (практически в 2 раза больше следующей марки). Также в топ лидеров входят "Опель", "БМВ", "Мерседес", "Ауди" и "Форд" 

5) большинство машин не было в ремонте (72,5%), однако еще про статус 17,5% машин нам неизвестно

6) даты в столбцах date_crawled и date_created почти идентичны.

**Feature Engineering** на основе признаков связанных с датой был создан признак 'relevance' (разница между 'last_seen' и 'date_crawled')-  актуальности объявления c момента скачивания из базы. Также объединенный признак brand_mod и при корреляционном анализе для устранения мультиколлинеарности мы объединили тип кузова и тип топлива в 'vehicle_fuel_type'.

**Корреляционный анализ** данных показал несколько мультиколлинеарных зависимостей. Поэтому нам пришлось избавиться от некоторых признаков или объединить их с другими. Построение попарных скаттерплотов не дало нам какой-то важной информации кроме того, что мощность имеет взаимосвязь с годом регистрации - старые авто маломощные.

С помощью **phik-корреляции** были высчитаны следующие коэффмциенты зависимости:
наиболее важные фичи brand_mod (0.58), registration_year(0.56), power (0.52). Достаточно хорошие repaired (0.35), vehicle_fuel_type(0.32), kilometer(0.31), gearbox(0.29). Слабые признаки registration_month (0.15), relevance(0.13), postal_code(0.09)

Т.к. подбор гиперпараметров может был очень долгим и трудоемким, для того, чтобы выкинуть ненужные признаки мы построили **baseline model** CatBoostRegressor(iterations=1000, learning_rate=0.05, depth=10), посмотрели ее Feature Importance на основе полученного скоринга и удалили два самых слабых признака: registration_month 1.45 и postal_code 1.32. 

**Выбор модели и подбор гиперпараметров**. Далее перебрали разные модели несколькими пайплайнами. Модель линейной регрессии показала очень плохие результаты и большое время обучения.

Перебор гиперпараметров для модели **LGBMRegressor** показал хорошие результаты (Кросс-валидация: RMSE для лучшей модели: 1627,кросс-валидация: MSE для лучшей модели: 26468297,кросс-валидация: MAE для лучшей модели: 10207. Время обучения: 13.2 секунды, время предикта: 7.3 секунды)

Аналогично подобрали гиперпараметры для модели катбуста. Скоринг оказался близким к нашей baseline модели (Кросс-валидация: RMSE для лучшей модели: 1713, кросс-валидация: MSE для лучшей модели: 2933331, кросс-валидация: MAE для лучшей модели: 1084).
